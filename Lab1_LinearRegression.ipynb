# Linear Regression

Let's fabricate some data that shows a relationship between page speed and amount purchased:

```python
%matplotlib inline
import numpy as np
from pylab import *

pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000)) * 3

scatter(pageSpeeds, purchaseAmount)
```

As we only have two features, we can keep it simple and just use scipy.state.linregress:

```python
from scipy import stats

slope, intercept, r_value, p_value , std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

The R-squared value shows the following fit:

```python
r_value ** 2
```

Let's use the slope and intercept we got from the regression to plot predicted values vs. observed:

```python
import matplotlib.pyplot as plt

def predict(x):
    return slope * x + intercept

fitLine = predict(pageSpeeds)

plt.scatter(pageSpeeds, purchaseAmount)
plt.plot(pageSpeeds, fitLine, c='r')
plt.show()
```

The function `stats.linregress` from **SciPy** is mainly designed to compute:

- **slope**  
- **intercept**  
- **r-value** (correlation coefficient)  
- **p-value**  
- **standard error of the slope**

It does **not** directly compute **RMSE** — but you can easily calculate it yourself using the predictions it gives you.

```python
# Predicted values
y_pred = intercept + slope * pageSpeeds

# RMSE
rmse = np.sqrt(np.mean((purchaseAmount - y_pred)**2))

print("RMSE:", rmse)
print("R²:", r_value**2)
```

## Activity

```python
# Compute evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Predicted values
y_pred = intercept + slope * pageSpeeds

# MAE
mae = mean_absolute_error(purchaseAmount, y_pred)

# MSE
mse = mean_squared_error(purchaseAmount, y_pred)

# RSS (Residual Sum of Squares)
rss = np.sum((purchaseAmount - y_pred) ** 2)

# RAE (Relative Absolute Error)
mean_actual = np.mean(purchaseAmount)
rae = mae / np.sum(np.abs(purchaseAmount - mean_actual))

# R2 and RMSE
r2 = r_value ** 2
rmse = np.sqrt(mse)

print("MAE:", mae)
print("MSE:", mse)
print("RSS:", rss)
print("RAE:", rae)
print("R2:", r2)
print("RMSE:", rmse)
```

```python
# Effect of increased random variation
print("Effect of increased random variation on R2:")
import numpy as np
from scipy import stats

# Generate data with higher variation
pageSpeeds_high_var = np.random.normal(3.0, 1.0, 1000)
noise = np.random.normal(0, 1.0, 1000)  # Higher noise
purchaseAmount_high_var = 100 - (pageSpeeds_high_var + noise) * 3

# Regress
slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds_high_var, purchaseAmount_high_var)
r2_high = r_value ** 2

print("Original R2: 0.9897")
print(f"High variation R2: {r2_high:.4f}")
print("As random variation increases, R2 decreases (model explains less variance)")
```
