# Polynomial Regression

What if your data doesn't look linear at all? Let's look at some more realistic-looking page speed / purchase data:

```python
%matplotlib inline
from pylab import *
import numpy as np

#seed to generate the same random numbers on the different machines
np.random.seed(2)
pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmount = np.random.normal(50.0, 10.0, 1000) / pageSpeeds

scatter(pageSpeeds, purchaseAmount)
```

numpy has a handy polyfit function we can use, to let us construct an nth-degree polynomial model of our data that minimizes squared error. Let's try it with a 4th degree polynomial:

```python
x = np.array(pageSpeeds)
y = np.array(purchaseAmount)

#Fit a polynomial of degree d=4 to points (x, y). Returns a vector of coefficients that minimises the squared error
p4 = np.poly1d(np.polyfit(x, y, 4))
```

We'll visualize our original scatter plot, together with a plot of our predicted values using the polynomial for page speed times ranging from 0-7 seconds:

```python
import matplotlib.pyplot as plt

# linspace returns evenly spaced numbers over a specified interval
xp = np.linspace(0, 7)
plt.scatter(x, y)
plt.plot(xp, p4(xp), c='r')
plt.show()
```

Measure the r-squared error :

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, p4(x))

print(r2)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

# 1. Predict values
y_pred = p4(x)

# 2. Compute metrics
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
```

## Activity

- Try different polynomial orders smaller and bigger than 4. Can you get a better fit with higher orders? What are your comments on the shape of the plot and the R-sqruared measure?

```python
# Try different polynomial orders smaller and bigger than 4
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

degrees = [2, 3, 4, 5, 6, 7, 8]
results = []
for d in degrees:
    p = np.poly1d(np.polyfit(x, y, d))
    y_pred = p(x)
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    r2 = r2_score(y, y_pred)
    results.append((d, rmse, r2))
    print(f"Degree {d}: RMSE={rmse:.4f}, R2={r2:.4f}")

# Find best degree (lowest RMSE)
best = min(results, key=lambda x: x[1])
print(f"\nBest degree: {best[0]} with RMSE={best[1]:.4f} and R2={best[2]:.4f}")

# Plot the best-fitting polynomial
best_degree = best[0]
p_best = np.poly1d(np.polyfit(x, y, best_degree))
import matplotlib.pyplot as plt
xp = np.linspace(0, 7, 100)
plt.scatter(x, y)
plt.plot(xp, p_best(xp), c='r')
plt.title(f"Best fit: degree {best_degree}")
plt.show()
```

```python
# Activity: Which degree works best?
print("Based on the results above:")
- Lower degree polynomials (2-3) underfit the data
- Degree 4 (original) provides moderate fit
- Higher degrees (5-8) progressively reduce RMSE but may overfit
- The best degree depends on the bias-variance tradeoff:
  * Higher degrees reduce training error but increase test error (overfitting)
  * Lower degrees give more robust models at cost of fit quality

print("Comments on shape and R-squared:")
The data appears to have a nonlinear relationship (purchaseAmount ~ 50/speed),
so higher-degree polynomials capture this curvature better, but degree 8 may
overfit the specific training data sampled.
```
